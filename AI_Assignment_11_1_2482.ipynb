## Task 1
**Prompt:** Generate a Stack class with push, pop, peek, and is_empty.

```python
class Stack:
    def __init__(self):
        self.items=[]

    def push(self,item):
        self.items.append(item)

    def pop(self):
        return self.items.pop() if self.items else 'Stack Empty'

    def peek(self):
        return self.items[-1] if self.items else None

    def is_empty(self):
        return len(self.items)==0

s=Stack()
s.push(10)
s.push(20)
print(s.pop())
```

**Observation:** Stack follows LIFO principle where the last inserted element is removed first.

## Task 2
**Prompt:** Implement a Queue using Python list.

```python
class Queue:
    def __init__(self):
        self.items=[]

    def enqueue(self,x):
        self.items.append(x)

    def dequeue(self):
        return self.items.pop(0)

q=Queue()
q.enqueue(1)
q.enqueue(2)
print(q.dequeue())
```

**Observation:** Queue follows FIFO principle where the first inserted element is processed first.
