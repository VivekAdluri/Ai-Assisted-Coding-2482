## Task 1
**Prompt:** Implement Bubble Sort to rank student scores.

```python
scores=[78,45,90,62,88]
n=len(scores)

for i in range(n):
    swapped=False
    for j in range(0,n-i-1):
        if scores[j]>scores[j+1]:
            scores[j],scores[j+1]=scores[j+1],scores[j]
            swapped=True
    if not swapped:
        break

print('Sorted Scores:',scores)
```

**Observation:** Bubble Sort repeatedly compares adjacent elements and swaps them to achieve sorted order.

## Task 2
**Prompt:** Implement Linear Search to find a student roll number.

```python
def linear_search(arr,target):
    for i,v in enumerate(arr):
        if v==target:
            return i
    return -1

print(linear_search([101,102,103,104],103))
```

**Observation:** Linear Search scans elements sequentially until the target is found.
