## Task 1
**Prompt:** Refactor legacy Python code using a more Pythonic approach such as list comprehension.

```python
values = [2,4,6,8,10]
doubled = [v*2 for v in values]
print(doubled)
```

**Observation:** List comprehension performs the transformation in one line, improving readability and efficiency.

## Task 3
**Prompt:** Use a safer dictionary access method.

```python
config={'host':'localhost','port':8080}
timeout=config.get('timeout')
if timeout:
    print(timeout)
else:
    print('Default timeout used')
```

**Observation:** `dict.get()` safely retrieves values without raising errors if the key is missing.
